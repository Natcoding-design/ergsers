```python
import pandas as pd
import numpy as np

np.random.seed(42)

n_users = 5000

df = pd.DataFrame({
    "user_id": range(1, n_users + 1),
    "group": np.random.choice(["A", "B"], size=n_users),
})

# вероятность конверсии (B чуть лучше)
conversion_rates = {"A": 0.10, "B": 0.12}

df["converted"] = df["group"].apply(
    lambda x: np.random.binomial(1, conversion_rates[x])
)

# добавим дополнительные метрики
df["revenue"] = df["converted"] * np.random.gamma(5, 10, size=n_users)

df["device"] = np.random.choice(
    ["mobile", "desktop"], size=n_users, p=[0.7, 0.3]
)

df["country"] = np.random.choice(
    ["LV", "EE", "LT"], size=n_users, p=[0.5, 0.3, 0.2]
)

df["date"] = pd.to_datetime("2025-01-01") + pd.to_timedelta(
    np.random.randint(0, 30, size=n_users), unit="D"
)

df.head(50)
```

```python
import numpy as np
from scipy import stats

# Данные: веса студентов в кг
data = np.array([68, 72, 70, 69, 71, 73, 70, 68, 72, 69])

# Параметры генеральной совокупности
mu = 70        # гипотетическое среднее
sigma = 2.5    # известное стандартное отклонение

# Размер выборки
n = len(data)

# Вычисляем Z-статистику
z_score = (np.mean(data) - mu) / (sigma / np.sqrt(n))

# Вычисляем p-value (двусторонний тест)
p_value = 2 * (1 - stats.norm.cdf(abs(z_score)))

print("Z-статистика:", z_score)
print("p-value:", p_value)
```
